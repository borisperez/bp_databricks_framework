This example notebook helps you to debug a specific Function configured in your task file.
If you want to test multiple chained functions one by one, you'd better to run this notebook
in Databricks because local run can only do "run all".

```python
from admcore.tools.debugger.db_debugger import AdmDebugger

debugger = AdmDebugger(
    databricks_path='/Workspace/Users/boris.bperez@gmail.com/.bundle/bp_databricks_framework/dev/files',
    project_root_path='src',
    env_file_path='src/env.toml',
    pipeline_file_path='src/test_bp_framework/pipeline.yml'
)
```

```python
output_df_func1 = debugger.debug_function(
    input_data=None,
    task_name='',       # Task key in pipeline.yml file
    action_name='',     # Action name in task config file
    function_name=''    # Function name in this Action
)

# Convert small chunk of Spark dataframe to Pandas for better display
output_df_func1.limit(10).toPandas()
```

```python
output_df_func2 = debugger.debug_function(
    input_data=output_df_func1,
    task_name='',       # Task key in pipeline.yml file
    action_name='',     # Action name in task config file
    function_name=''    # Function name in this Action
)

# Convert small chunk of Spark dataframe to Pandas for better display
output_df_func2.limit(10).toPandas()
```
